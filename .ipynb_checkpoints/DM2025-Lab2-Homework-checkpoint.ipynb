# **Project Report**

## 1. Model Development (10 pts Required)

### 1.1 Preprocessing Steps

First I examined the data and found that it was already quite clean.
The raw posts are stored in `final_posts.json`, the labels are in `emotion.csv`, and `data_identification.csv` indicates which `id`s belong to the training set and which belong to the test set.
I transform the JSON data into a pandas `DataFrame` and merge on `id` with `data_identification.csv` and `emotion.csv`.

Then I implement two functions, `normalize_text` and `process_hashtags`, to preprocess the texts.

#### `process_hashtags`

Since the original hashtag data is a list and has no `#`, in this function I combine them into a single string such as `#tag1 #tag2` and add it to the post text in the `normalize_text` function.

#### `normalize_text`

I utilize regular expressions to match and replace placeholders and URLs with new tokens.

* URL handling

  * Replace URLs with the `<URL>` token.

* Emoji handling

  * Use `emoji.demojize` to convert emojis to `EMO_{emoji_name}`.

* Name and user handling

  * Replace `[Name]` with the `<NAME>` token.
  * Replace `@user` with the `<USER>` token.

### 1.2 Feature Engineering Steps

I did not create new non-text features; I only performed feature engineering on the text.

* Token replacement

  * As previously stated in the preprocessing steps, URLs, `[Name]`, and `@user` are replaced with new tokens. This helps the model learn them as single tokens instead of splitting them into multiple different tokens.
  * I add these new tokens to the tokenizer vocabulary list.

* Hashtag integration

  * I concatenate the hashtags with the post text, since hashtags sometimes contain strong emotional information.

* Emoji encoding

  * I convert emojis to `EMO_{emoji_name}` to add semantic information while maintaining the `EMO_` prefix, which tells the model it is an emoji and not regular text content.

### 1.3 Explanation of Your Model

I fine-tune a pretrained model to generate pseudo labels, combine the original training data and pseudo-labeled data, and then fine-tune the final model.

* Pretrained model fine-tuning

  * I fine-tune `roberta-base` for a small number of epochs (e.g., 2) and use it to generate pseudo labels.

* Pseudo labels

  * I select pseudo labels for which the model is confident enough, defined as maximum class probability greater than or equal to 0.9.
  * I also require the uncertainty to be low enough, defined as entropy less than or equal to 0.2.
  * To ensure that the number of pseudo labels is not too large and does not corrupt the final result, I cap the number of pseudo-labeled examples at 10 percent of the test data size; if there are more, I randomly sample from them.

* Final model

  * I fine-tune the model again on both the pseudo-labeled data and the original training data for more epochs (e.g., 3).

# **Competition Code**

## 1. Preprocessing Steps

```python
import os,json,re,emoji
import numpy as np
import pandas as pd
import torch
# import torch.nn.functional as F

from scipy.stats import entropy
from scipy.special import softmax
from datasets import Dataset
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score, confusion_matrix, classification_report

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
)
```

```python
seed = 42
def set_seed(seed=42):
    np.random.seed(seed)
    torch.manual_seed(seed)
set_seed(seed)
```

```python
MODEL_NAME = "roberta-base"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
```

```python
DIR_PATH = "dm-lab-2-private-competition/"
POSTS_PATH = f"{DIR_PATH}final_posts.json"
EMOTION_PATH = f"{DIR_PATH}emotion.csv"
DATA_ID_PATH = f"{DIR_PATH}data_identification.csv"
SUB_PATH = f"{DIR_PATH}samplesubmission.csv"

emotion_df = pd.read_csv(EMOTION_PATH)
id_df = pd.read_csv(DATA_ID_PATH)
```

```python
with open(POSTS_PATH,"r",encoding="utf-8") as f:
    raw_posts = json.load(f)
```

```python
# generate df from json
ids = []
txts = []
hashtgs = []
for i in raw_posts:
    root = i['root']
    post = root['_source']['post']
    ids.append(post['post_id'])
    txts.append(post['text'])
    hashtgs.append(post['hashtags'])
posts_df = pd.DataFrame({
    "id":ids,
    "text":txts,
    "hashtags":hashtgs
})
```

```python
df = posts_df.merge(id_df,on="id",how="inner").merge(emotion_df,on="id",how="left")
```

## 2. Feature Engineering Steps

```python
def process_hashtags(tags:list)->str:
    ntags = tags
    l = len(tags)
    for i in range(l):
        tag = tags[i]
        TAG_PATTERN = re.compile(r"[A-Z]?[a-z]+|[A-Z]+(?=[A-Z]|$)")
        if tag.find("_") != -1:
            tag = " ".join(tag.split("_"))
        t = TAG_PATTERN.findall(tag)
        if t:
            
            tag = " ".join(i for i in t if i)
        ntags[i] = tag
    # add # back since original hashtag don't have
    tag_text = " ".join("#"+i for i in ntags if type(i)==str)
    return tag_text

# posts_df["tags"] = posts_df["hashtgs"].apply(split_hashtags)


def normalize_text(text, hashtags):
    # normalize text based on common pattern found in tweet using regx
    tag = process_hashtags(hashtags or "")
    txt = (text or "") + " " + tag
    # demojized turn emoji to EMO_{name}
    demojized = emoji.demojize(txt, language="en", delimiters=("EMO_"," "))
    st = demojized
    st = re.sub(r"http\S+|www\.\S+", " <URL> ", st) # url
    st = re.sub(r"@\w+", " <USER> ", st) # tag user
    st = re.sub(r"\s+", " ", st).strip() # multiple space \t \n or \r 
    st = re.sub(r"\[NAME\]","<NAME>",st, flags=re.IGNORECASE) # name flag
    st = re.sub(r"edit:","",st,flags=re.IGNORECASE)
    st = re.sub(r"tl;dr","",st,flags=re.IGNORECASE) 
    return st

df["text_clean"] = df.apply(
    lambda row: normalize_text(row["text"], row["hashtags"]), axis=1)

# update tokenizer
new_token = ["<URL>","<USER>","<NAME>"]
tokenizer.add_tokens(new_token)
# after preprocessing text spilt train and test
train_df = df[df["split"]=="train"]
test_df = df[df["split"]=="test"]
```

```python
# map emotion to number
labels = sorted(train_df["emotion"].unique())
label2id = {l: i for i, l in enumerate(labels)}
id2label = {i: l for l, i in label2id.items()}

print("Labels:", labels)
print("label2id:", label2id)
train_df["label"] = train_df["emotion"].map(label2id)
```

```python
def make_dataset(texts, labels=None):
    data = {"text": texts}
    if labels:
        data["label"] = labels
    ds = Dataset.from_dict(data)
    ds = ds.map(tokenize_fn, batched=True)
    ds = ds.remove_columns(["text"])
    ds.set_format("torch")
    return ds

def tokenize_fn(batch,max_length=256):
    return tokenizer(
        batch["text"],
        truncation=True,
        padding="max_length",
        max_length=max_length,
    )

def compute_metrics(eval_pred):
    logits, label_ids = eval_pred
    preds = np.argmax(logits, axis=-1)
    return {
        "f1_macro": f1_score(label_ids, preds, average="macro"),
        "f1_micro": f1_score(label_ids, preds, average="micro")
    }
```

## 3. Model Implementation Steps

```python
# generate full train dataset for pseudo label generation
full_texts = train_df["text_clean"].tolist()
full_labels = train_df["label"].tolist()

full_ds = make_dataset(full_texts, full_labels)
```

```python
# train model for  pseudo label
pseudo_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=len(labels),
    id2label=id2label,
    label2id=label2id,
)
pseudo_model.resize_token_embeddings(len(tokenizer))

pseudo_training_args = TrainingArguments(
    output_dir="./roberta-emotion-pseudo",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    num_train_epochs=2,
    logging_steps=100,
    save_strategy="epoch",
    metric_for_best_model="f1_micro"
)

pseudo_trainer = Trainer(
    model=pseudo_model,
    args=pseudo_training_args,
    train_dataset=full_ds,
    tokenizer=tokenizer,
    compute_metrics=compute_metrics,
)

pseudo_trainer.train()
print(pseudo_trainer.evaluate(full_ds))
```

```python
print(len(test_df),len(train_df))
```

```python
# create test dataset for pseudo label
test_texts = test_df["text_clean"].tolist()
test_ids = test_df["id"].tolist()
test_ds_for_pseudo = make_dataset(test_texts, labels=None)

pseudo_logits = pseudo_trainer.predict(test_ds_for_pseudo).predictions
pseudo_ids = np.argmax(pseudo_logits, axis=-1)
pseudo_label_names = [id2label[i] for i in pseudo_ids]
```

```python
pseudo_probs = softmax(pseudo_logits,axis=1)
max_probs = pseudo_probs.max(axis=1)
ent = entropy(pseudo_logits,axis=1)

threshold = 0.9

keep_mask = (max_probs >= threshold) & (ent<=0.2)


test_df["pseudo_label_id"] = pseudo_ids
test_df["pseudo_label_name"] = pseudo_label_names
max_pseudo = int(len(test_df)*0.1)
pseudo_df = test_df.loc[keep_mask]
if len(pseudo_df) > max_pseudo:
    pseudo_df =  pseudo_df.sample(max_pseudo, random_state=seed)

X_pseudo = pseudo_df["text_clean"]
y_pseudo = pseudo_df["pseudo_label_id"]
# test_df["pseudo_conf"] = max_probs
# test_df["use_for_pseudo"] = keep_mask

print("pseudo kept:  ", len(pseudo_df),"/",len(keep_mask),"threshold ",threshold)
print("pseudo label distribution:")
print(pseudo_df["pseudo_label_name"].value_counts())
print("pseudo label distribution (all):")
print(test_df["pseudo_label_name"].value_counts())
```

```python
# train using 3fold cv to validate the f1 score of adding pseudo label 
X = np.array(full_texts)
y_label = np.array(full_labels)

skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

train_f1_list = []
val_f1_list = []

for fold, (tr_idx, val_idx) in enumerate(skf.split(X, y_label), start=1):
    print(f"fold {fold}:")
    train_texts = X[tr_idx].tolist()
    train_labels = y_label[tr_idx].tolist()
    val_texts = X[val_idx].tolist()
    val_labels = y_label[val_idx].tolist()

    train_comb_texts = train_texts + X_pseudo.tolist()
    train_comb_labels = train_labels  + y_pseudo.tolist()

    print(f"fold {fold} true:{len(train_texts)}+pseudo: {len(X_pseudo)} total:{len(train_comb_texts)}")
    print(f"fold {fold} val_size:{len(val_texts)}")

    ds_tr = make_dataset(train_comb_texts, train_comb_labels)
    ds_val = make_dataset(val_texts, val_labels)

    cv_model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=len(labels),
        id2label=id2label,
        label2id=label2id,
    )
    cv_model.resize_token_embeddings(len(tokenizer))
    cv_args = TrainingArguments(
        output_dir=f"./roberta-emotion-cv-fold{fold}",
        learning_rate=2e-5,
        per_device_train_batch_size=16,
        per_device_eval_batch_size=32,
        num_train_epochs=2,
        logging_steps=100,
        save_steps=1000,
        eval_strategy="epoch",
        metric_for_best_model="f1_micro"
    )

    cv_trainer = Trainer(
        model=cv_model,
        args=cv_args,
        train_dataset=ds_tr,
        eval_dataset=ds_val,
        tokenizer=tokenizer,
        compute_metrics=compute_metrics,
    )

    print(f"Fold {fold} Training")
    cv_trainer.train()

    print(f"Fold {fold} Evaluating on validation fold")
    eval_res = cv_trainer.evaluate(ds_val)
    print(f"[Fold {fold}] Eval metrics:", eval_res)

    fold_f1_mac = eval_res.get("eval_f1_macro", eval_res.get("f1_macro"))
    fold_f1_mic = eval_res.get("eval_f1_micro", eval_res.get("f1_micro"))
    val_f1_list.append((fold_f1_mac,fold_f1_mic))
```

```python
print("3fold CV F1:", val_f1_list)

valid_mac_scores = [s[0] for s in val_f1_list if s is not None]
if valid_mac_scores:
    print("mean F1-macro:", np.mean(valid_mac_scores))
    print("std F1-macro:", np.std(valid_mac_scores))


valid_mic_scores = [s[1] for s in val_f1_list if s is not None]
if valid_mic_scores:
    print("mean F1-micro:", np.mean(valid_mic_scores))
    print("std F1-micro:", np.std(valid_mic_scores))
```

```python
print("train final model on all train split + high confiednce psuedo label")

combined_texts = full_texts + X_pseudo.tolist()
combined_labels = full_labels + y_pseudo.tolist()
combined_ds = make_dataset(combined_texts, combined_labels)


final_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=len(labels),
    id2label=id2label,
    label2id=label2id,
)
final_model.resize_token_embeddings(len(tokenizer))

final_args = TrainingArguments(
    output_dir="./roberta-emotion-final",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    num_train_epochs=3,
    logging_steps=100,
    save_steps=1000,
    metric_for_best_model="f1_micro"
)

final_trainer = Trainer(
    model=final_model,
    args=final_args,
    train_dataset=combined_ds,
    tokenizer=tokenizer,
    compute_metrics=compute_metrics,
)

final_trainer.train()
print(final_trainer.evaluate(combined_ds))
```

```python
final_test_ds = make_dataset(test_texts, labels=None)
final_logits = final_trainer.predict(final_test_ds).predictions
final_pred_ids = np.argmax(final_logits, axis=-1)
final_pred_labels = [id2label[i] for i in final_pred_ids]

submission = pd.DataFrame({
    "id": test_ids,
    "emotion": final_pred_labels,
})

submission_path = "submission_3cv.csv"
submission.to_csv(submission_path, index=False)
print("saved to:", submission_path)
display(submission.head())
```

```python
final_logits = final_trainer.predict(full_ds).predictions
final_pred = np.argmax(final_logits, axis=-1)
display(confusion_matrix(full_labels, final_pred))
print(classification_report(full_labels, final_pred))
```
